### 27. Add and replace `Code` in `travel_df`

```python
travel_df = pd.DataFrame({
    'Traveler': ['A', 'B', 'C', 'D', 'E'],
    'Destination': ['Delhi', 'Goa', 'Jaipur', 'Kochi', 'Manali'],
    'Days': [2, 4, 3, 5, 6]
})

travel_df['Code'] = np.arange(5)
print('After assigning np.arange(5):')
print(travel_df)

travel_codes = pd.Series(['TR-101', 'TR-104'], index=[1, 4], name='Code')
travel_df['Code'] = travel_codes
print('After assigning the partially indexed Series:')
print(travel_df)
```

### 28. Add and fill the missing `House` column

### 29. Add a partially indexed Series and delete it

```python
inventory_df = pd.DataFrame({
    'Item': ['Pen', 'Notebook', 'Pencil', 'Eraser'],
    'Quantity': [20, 15, 30, 10],
    'Price': [10, 50, 5, 8]
})

stock_locations = pd.Series(['Shelf A', 'Shelf C'], index=[0, 2], name='Location')
inventory_df['Location'] = stock_locations
print('After adding Location:')
print(inventory_df)

inventory_df = inventory_df.drop(columns='Location')
print('After deleting Location:')
print(inventory_df)
```

### 30. Complete sequence using `lab_df`

```python
# Copy a table with at least four columns before running this cell.
lab_df = pd.read_clipboard()

print('Complete DataFrame:')
print(lab_df)

print('Column names:')
print(lab_df.columns)

# Retrieve the first single-word column.
single_word_columns = [col for col in lab_df.columns if ' ' not in str(col)]
if single_word_columns:
    print('Single-word column:')
    print(lab_df[single_word_columns[0]])

# Retrieve the first column whose name contains a space.
space_columns = [col for col in lab_df.columns if ' ' in str(col)]
if space_columns:
    print('Column containing a space:')
    print(lab_df[space_columns[0]])

# Construct a smaller DataFrame from the first three available columns.
selected_columns = list(lab_df.columns[:3])
small_lab_df = lab_df[selected_columns]
print('Smaller DataFrame:')
print(small_lab_df)

# Add and then delete a whole column.
lab_df['PracticeColumn'] = 'Added'
print('After adding PracticeColumn:')
print(lab_df)

lab_df = lab_df.drop(columns='PracticeColumn')
print('After deleting PracticeColumn:')
print(lab_df)
```
