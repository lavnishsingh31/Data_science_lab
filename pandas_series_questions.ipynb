## Section C – Hard

### 26. Addition aligns matching index labels

```python
north = Series({'Jan': 40, 'Feb': 55, 'Mar': 60})
south = Series({'Feb': 20, 'Mar': 25, 'Apr': 30})
total = north + south
print(total)
# Feb = 55 + 20 = 75, Mar = 60 + 25 = 85; Jan and Apr become NaN.
```

### 27. Missing data after creating a Series with a custom index

```python

d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green', 'Red', 'Yellow', 'Blue']
result = Series(d, index=labels)
print(result)
print('\n(a) NaN label:', result[result.isnull()].index[0])
print('(b) isnull for Yellow:', pd.isnull(result)['Yellow'])
print('(c) Yellow is absent from the dictionary.')
print('(d) Numeric labels:', result[result.notnull()].index.tolist())
```

### 28. Convert a Series to dictionary and back

```python

s = Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])
d = s.to_dict()
s2 = Series(d)
print('Original Series:')
print(s)
print('\nReconstructed Series:')
print(s2)
print('\nIndex:', s2.index.tolist())
print('Values:', s2.values.tolist())
print('Same labelled data:', s.equals(s2))
```

### 29. Labelled product sales program

```python

products = Series([250, 600, 450, 800], index=['P1', 'P2', 'P3', 'P4'])
print('Product sales:')
print(products)
print('\nValues:', products.values)
print('Index:', products.index)
print('Selected P2:', products['P2'])
print('\nSales above 500:')
print(products[products > 500])
print("\nDoes P3 exist?", 'P3' in products.index)
print('Dictionary:', products.to_dict())
```

### 30. Analyse index order, missing data, filtering, names

```python

data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'

print(s)
print('\n(a) Index order:', s.index.tolist())
print('\n(b) NaN label:', s[s.isnull()].index[0])
print('\n(c) Values greater than 20:')
print(s[s > 20])
print('\n(d) pd.notnull(s):')
print(pd.notnull(s))
print('\n(e) Series name:', s.name)
print('Index name:', s.index.name)
```
